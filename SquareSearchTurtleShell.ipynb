```python
def get_kxk_nearest_signed(cx, cy, k, min_val=-1024, max_val=1016):
    r = k // 2

    # domain limits
    x_min = cx - r - 1
    x_max = cx + r 
    y_min = cy - r - 1
    y_max = cy + r

    # boundary shift 
    if x_min < min_val:
        shift = min_val - x_min
        x_min += shift
        x_max += shift
    elif x_max > max_val:
        shift = x_max - max_val
        x_min -= shift
        x_max -= shift

    if y_min < min_val:
        shift = min_val - y_min
        y_min += shift
        y_max += shift
    elif y_max > max_val:
        shift = y_max - max_val
        y_min -= shift
        y_max -= shift

    if 0 in range(x_min, x_max):
        x_max = x_max + 1
    if 0 in range(y_min, y_max):
        y_max = y_max + 1
    
    neighbors = []
    for x in range(x_min, x_max):
        for y in range(y_min, y_max):
            # skip zero axis
            if x == 0 or y == 0: continue
            neighbors.append((x, y))

    return neighbors
```

```python
def testing():
    test_points = [(-1, -1), (1, 1), (-1, 1), (1, -1), (-1, -2), (1, 2), (-1, 2), (1, -2)]
    k = [3, 5]
    for size in k:
        for p in test_points:
            neighbor = get_kxk_nearest_signed(p[0], p[1], k=size)
            print(f"Point {p} with k={size} count={len(neighbor)} neighbors: {neighbor}")
        print("-----")
```
